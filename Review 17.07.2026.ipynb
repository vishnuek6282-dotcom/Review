# Section 3: Outlier Detection

## Q11 A11

```python
cols=["insulin_level","triglycerides","cholesterol_total","waist_to_hip_ratio"]

for col in cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

## Q12 A12

```python
Q1=df["bmi"].quantile(0.25)
Q3=df["bmi"].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[(df["bmi"]<lower)|(df["bmi"]>upper)]

print("Number of BMI Outliers:",len(outliers))
```
